```python
import sympy as sp

# Variables
x, y = sp.symbols('x y')

# ==== (a) Expansión genérica en bases monomiales ====
def inciso_a():
    base_x = [1, x, x**2]
    base_y = [1, y, y**2]

    a0, a1, a2, b0, b1, b2 = sp.symbols('a0 a1 a2 b0 b1 b2')
    p = a0 + a1*x + a2*x**2
    g = b0 + b1*y + b2*y**2
    tensor = sp.expand(p*g)

    coeffs = {}
    for i, bx in enumerate(base_x):
        for j, by in enumerate(base_y):
            coeffs[(i,j)] = sp.expand(tensor).coeff(bx*by)

    return tensor, coeffs


# ==== (b) Polinomios dados ====
def inciso_b():
    p = x**2 + x + 3
    g = y + 1
    return sp.expand(p*g)


# ==== (c) Expansión en Legendre ====
def inciso_c():
    P0 = sp.legendre(0, x)
    P1 = sp.legendre(1, x)
    P2 = sp.legendre(2, x)
    base_legendre = [P0, P1, P2]

    p = x**2 + x + 3
    c0, c1, c2 = sp.symbols('c0 c1 c2')
    expr = c0*P0 + c1*P1 + c2*P2

    sol = sp.solve(sp.Eq(p, expr), (c0, c1, c2))
    return base_legendre, sol


# ==== (d) Tensor en Legendre ====
def inciso_d():
    Px = [sp.legendre(i, x) for i in range(3)]
    Py = [sp.legendre(i, y) for i in range(3)]

    p = x**2 + x + 3
    g = y + 1
    tensor = sp.expand(p*g)

    coeffs = {}
    for i, bx in enumerate(Px):
        for j, by in enumerate(Py):
            coeffs[(i,j)] = sp.expand(tensor).coeff(bx*by)

    return tensor, coeffs


# ================== MAIN ===================
if __name__ == "__main__":
    # (a)
    print("\n=== Inciso (a): Tensor genérico en bases monomiales ===\n")
    tensor_a, coeffs_a = inciso_a()
    print(f"f(x,y) = {sp.latex(tensor_a)}")
    print("Coeficientes c^{ij}:")
    for (i,j), val in coeffs_a.items():
        if val != 0:
            print(f"  c^{{{i}{j}}} = {sp.latex(val)}")

    # (b)
    print("\n=== Inciso (b): Con p^P(x)=x²+x+3 y p^G(y)=y+1 ===\n")
    print(f"f(x,y) = {sp.latex(inciso_b())}")

    # (c)
    print("\n=== Inciso (c): Expansión en base de Legendre ===\n")
    base_legendre, coeffs_c = inciso_c()
    for i, Pi in enumerate(base_legendre):
        print(f"P_{i}(x) = {sp.latex(Pi)}")
    print("\nCoeficientes de la expansión:")
    for k,v in coeffs_c.items():
        print(f"  {sp.latex(k)} = {sp.latex(v)}")

    # (d)
    print("\n=== Inciso (d): Tensor en base de Legendre ===\n")
    tensor_d, coeffs_d = inciso_d()
    print(f"f(x,y) = {sp.latex(tensor_d)}")
    print("Coeficientes c^{ij}:")
    for (i,j), val in coeffs_d.items():
        if val != 0:
            print(f"  c^{{{i}{j}}} = {sp.latex(val)}")
```
